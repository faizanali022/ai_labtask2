```python
from collections import deque

# Defining a sample graph representation with connection weights
graph = {
    "Home": {"Park": 2, "Market": 4},
    "Park": {"Home": 2, "Library": 5, "Market": 1},
    "Market": {"Home": 4, "Park": 1, "Library": 2},
    "Library": {"Park": 5, "Market": 2}
}

def bfs_all_paths(graph, start, goal):
    queue = deque()
    queue.append((start, [start], 0))
    all_paths = []
    while queue:
        node, path, cost = queue.popleft()
        if node == goal:
            all_paths.append((path, cost))
        else:
            for neighbor, weight in graph[node].items():
                if neighbor not in path:
                    new_path = path + [neighbor]
                    new_cost = cost + weight
                    queue.append((neighbor, new_path, new_cost))
    return all_paths

all_paths = bfs_all_paths(graph, "Home", "Library")
for i, (path, cost) in enumerate(all_paths, 1):
    print(f"Path {i}: {path}, Cost:{cost}")
shortest = min(all_paths, key=lambda x: x[1])
print("shortest Path:", shortest[0], "Cost: ", shortest[1])
```

```python
def dfs_all_paths(graph, current, goal, path=None, cost=0):
  if path is None:
    path = []
  path = path + [current]

  if current == goal:
    return [(path, cost)]

  all_paths = []
  for neighbor, weight in graph[current].items():
    if neighbor not in path:
      new_paths = dfs_all_paths(graph, neighbor, goal, path, cost + weight)
      all_paths.extend(new_paths)
  return all_paths

all_paths = dfs_all_paths(graph, "Home", "Library")
for i, (path, cost) in enumerate(all_paths, 1):
  print(f"Path {i}: {path} , Cost: {cost}")
shortest = min(all_paths, key=lambda x: x[1])
print("shortest Path:", shortest[0], "Cost: ", shortest[1])
```
